```python
import pandas as pd

# Criando um DataFrame com um campo float
df = pd.DataFrame({'valor': [1234.567, 987.001, 55.22]})

# Convertendo para string
df['valor'] = df['valor'].astype(str)

# Formatando para duas casas decimais e separador de milhares
df['valor'] = df['valor'].str.replace('.', ',', regex=False)  # Substitui ponto por vírgula
df['valor'] = df['valor'].str.replace(r'(\d)(?=(\d{3})+(?!\d))', r'\1.', regex=True) # Adiciona separador de milhar
#df['valor'] = df['valor'].apply(lambda x: f'{float(x.replace(".", "").replace(",", ".")):.2f}')  # Converte de volta para float para formatar

# Exibindo o resultado
print(df.info())
print(df)
```

```python
# Para formatar como moeda (por exemplo, R$ 1.234,56)
def format_moeda(valor):
    return "{:.2f}".format(valor).replace('.', ',').replace(',', '.').replace(' ', '')  # Formata para duas casas decimais e troca ponto por vírgula (opcional)
```

```python
df['valor_formatado'] = df['valor'].apply(format_moeda)
```

```python
import pandas as pd
from decimal import Decimal

# Criar um DataFrame de exemplo
data = {'valor': [1234.567, 890.12, 500.0]}
df = pd.DataFrame(data)

# Converter para Decimal
df['valor'] = df['valor'].apply(lambda x: Decimal(str(x)))

# Formatar para moeda
def format_moeda(valor):
    return "{:.2f}".format(valor).replace('.', ',').replace(',', '.').replace(' ', '')  # Formata para duas casas decimais e troca ponto por vírgula (opcional)

df['valor_formatado'] = df['valor'].apply(format_moeda)

print(df)
```
